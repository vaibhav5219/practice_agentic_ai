```python
###  Open AI API key and open source models -- Llama3, Gemma2, Mistral --Groq etc. ###
import os
from dotenv import load_dotenv
load_dotenv("../../.env")
```

```python
import openai
openai.api_key = os.getenv("OPENAI_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq

model=ChatGroq(model="llama-3.1-8b-instant", groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

messages = [
    SystemMessage(content="Translate the following sentence from English to French."),
    HumanMessage(content="Helllo, how are you?"),
]

result=model.invoke(messages)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parser.invoke(result)
```

```python
### Using LCEL - chain the components
chain = model | parser
chain.invoke(messages) ## means the output of model will be in string only
```

```python
### Prompt Template
from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the following sentence into {language}:"
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", generic_template),
        ("user", "{text}")
    ]
)
```

```python
result=prompt.invoke({"language":"French", "text":"Hello"})
```

```python
result.to_messages()
```

```python
chain=prompt | model | parser
result=chain.invoke({"language":"Bhojpuri", "text":"Hello"})
print(result)
```
