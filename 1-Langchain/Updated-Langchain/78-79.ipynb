### Streaming And Batch

### Streaming
Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.

Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
# streaming
model.stream("Write me a poem about a cat in 500 words").for_each(print)
```

```python
# returns after whole text generates, not streaming
model.invoke("Hello, world!")
```

```python
for chunk in model.stream("Write me a poem about a cat in 500 words"):
    print(chunk.text, end="|", flush=True)
```


#### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
# many request we can pass in parallel and get response in list (parallel processing)
response = model.batch(
    [
        "Write me a poem about a cat in 500 words",
        "Write me a poem about a dog in 500 words",
    ]
)

for res in response:
    print(res.content)
```

```python
# many request we can pass in parallel and get response in list (parallel processing)
response = model.batch(
    [
        "Write me a poem about a cat in 500 words", # at a time only 2 request will process, rest will wait in queue
        "Write me a poem about a dog in 500 words",
        "Write me a poem about a rabbit in 500 words",
        "Write me a poem about a lion in 500 words",
    ],
    max_concurrency=2,  # Set the maximum number of concurrent requests (at a time only 2 request will process)
)

for res in response:
    print(res.content)
```
