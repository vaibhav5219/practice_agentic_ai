```python
from langchain.agents import create_agent

def get_weather(location: str) -> str:
    ''' This is a mock function that simulates getting the weather for a location.'''
    return f"The weather in {location} is sunny with a high of 25 degrees Celsius."

agent = create_agent(
    model = "gpt-3.5-turbo",
    system_prompt="""You are a helpful assistant that can use tools to answer questions.""",
    tools=[get_weather]
)
agent
```

```python
### Run the agent
response = agent.invoke({ "messages": [{ "role": "user", "content":"What is the weather in Delhi?"}] })
```
