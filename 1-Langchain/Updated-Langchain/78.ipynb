# Models Integration with OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
print(model)
```

```python
## invoke model
response=model.invoke("What is the capital of France?")
```

```python
response.content
```

### Google Gemini Model Integration

```python
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-3.1-pro-preview")
response = model.invoke("Why do parrot talk?")
response.content
```

```python
### ChatOpenAI
# Another Way

from langchain_openai import ChatOpenAI

model=ChatOpenAI(model="gpt-4.1")
response=model.invoke("Why do parrot talk?")
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.1-pro-preview")
response = model.invoke("Why do parrot talk?")
response
```
