## Pydentic Data Validation

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
from random import random
from pydantic import BaseModel, Field

class State(BaseModel):
    name: str = Field(..., description="The name of the player")
```

```python
## node functions
def example_state_node(state: State):
    return {"name": "Hello"}
```

```python
## StateGraph construction
builder = StateGraph(State)
builder.add_node("example_state_node", example_state_node)

builder.add_edge(START, "example_state_node")
builder.add_edge("example_state_node", END)

graph = builder.compile()
graph.invoke(State(name="Alice"))
```

```python
graph.invoke(State(name=123))
```
